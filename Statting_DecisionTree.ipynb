```python
# SERVE RECEIVE STATISTICS


# Player A: Pass Rating = (weighted average of all SR attempts)
A_PASS_SCORE = ((A_SR_0 * 0) + (A_SR_1 * 1) + (A_SR_2 * 2) + (A_SR_3 * 3)) / A_SR_TA

# Player A: MTP % = (swings - errors) / (total SR attempts - partner A2 attempts)
A_SR_MTP = (A_SR_A - A_SR_HE) / (A_SR_TA - B_SR_A2A)

# Player A: 3rd Ball FBSO % = (FB kills) / (total SR attempts - partner A2 attempts)
A_FBSO = A_SR_K / (A_SR_TA - B_SR_A2A) # likelihood of scoring when served

# Player A: FB Kill Rate = (FB kills + A2 kills) / (swings + A2 attempts)
A_FBK = (A_SR_K + A_SR_A2K) / (A_SR_A + A_SR_A2A) # likelihood of scoring when attacking

# Player A: SR A2 Efficiency = (A2 Kills - A2 Errors) / A2 Attempts
A_SRA2 = (A_SR_A2K - A_SR_A2E) / A_SR_A2A

# Player A: SR Attempt % = (share of team)
A_SR = A_SR_TA / (A_SR_TA + B_SR_TA)





# Player B: Pass Rating (weighted average of all SR attempts)
B_PASS_SCORE = ((B_SR_0 * 0) + (B_SR_1 * 1) + (B_SR_2 * 2) + (B_SR_3 * 3)) / B_SR_TA

# Player B: MTP % = (swings - errors) / (total SR attempts - partner A2 attempts)
B_SR_MTP = (B_SR_A - B_SR_HE) / (B_SR_TA - A_SR_A2A)

# Player B: 3rd Ball FBSO % = (FB kills) / (total SR attempts - partner A2 attempts)
B_FBSO = B_SR_K / (B_SR_TA - A_SR_A2A) # likelihood of scoring when served

# Player B: FB Kill Rate = (FB kills + A2 kills) / (swings + A2 attempts)
B_FBK = (B_SR_K + B_SR_A2K) / (B_SR_A + B_SR_A2A) # likelihood of scoring when attacking

# Player B: SR A2 Efficiency = (A2 Kills - A2 Errors) / A2 Attempts
B_SRA2 = (B_SR_A2K - B_SR_A2E) / B_SR_A2A

# Player B: SR Attempt % = (share of team)
B_SR = B_SR_TA / (A_SR_TA + B_SR_TA)
```
